```python

import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Create sample job postings data
np.random.seed(42)

n_jobs = 500  # Create 500 job postings

# Job titles
job_titles = [
    'Data Analyst', 'Data Scientist', 'Business Analyst', 'ML Engineer', 
    'Data Engineer', 'Product Analyst', 'Research Analyst', 'Analytics Manager',
    'BI Developer', 'Data Analyst Junior', 'Senior Data Analyst', 'Marketing Analyst'
]

# Companies
companies = [
    'Tech Corp', 'AI Solutions', 'Finance Inc', 'StartupXYZ', 'DataFlow',
    'ProductCo', 'ResearchLab', 'MegaCorp', 'TechVision', 'CloudSystems',
    'InfoTech', 'Digital Solutions', 'Analytics Pro', 'DataFirst', 'SmartAnalytics'
]

# Locations (focus on Indian cities)
locations = ['Hyderabad', 'Bangalore', 'Mumbai', 'Delhi', 'Pune', 'Chennai', 'Gurgaon', 'Kolkata']

# Job types
job_types = ['Remote', 'Hybrid', 'Onsite']

# Experience levels
experience_levels = ['Entry', 'Mid', 'Senior', 'Lead']

# Skills pool
skills_pool = [
    'Python', 'SQL', 'Excel', 'Power BI', 'Tableau', 'Machine Learning',
    'TensorFlow', 'PyTorch', 'AWS', 'Azure', 'Google Cloud', 'R', 'JavaScript',
    'Docker', 'Kubernetes', 'Airflow', 'Spark', 'Hadoop', 'DAX', 'Visualization'
]

# Generate random data
data = {
    'job_id': range(1, n_jobs + 1),
    'job_title': np.random.choice(job_titles, n_jobs),
    'company': np.random.choice(companies, n_jobs),
    'location': np.random.choice(locations, n_jobs, p=[0.22, 0.30, 0.15, 0.12, 0.08, 0.05, 0.05, 0.03]),
    'salary_min': np.random.randint(300000, 800000, n_jobs),
    'salary_max': np.random.randint(800000, 2000000, n_jobs),
    'job_type': np.random.choice(job_types, n_jobs, p=[0.35, 0.35, 0.30]),
    'experience_level': np.random.choice(experience_levels, n_jobs, p=[0.30, 0.40, 0.25, 0.05]),
    'company_size': np.random.choice(['10-50', '50-200', '200-500', '500+'], n_jobs, p=[0.25, 0.35, 0.25, 0.15]),
}

# Create DataFrame
df = pd.DataFrame(data)

# Generate random skills for each job (3-6 skills per job)
def generate_skills():
    num_skills = np.random.randint(3, 7)
    selected_skills = np.random.choice(skills_pool, num_skills, replace=False)
    return ', '.join(selected_skills)

df['required_skills'] = df.apply(lambda x: generate_skills(), axis=1)

# Generate posted dates (last 90 days)
base_date = datetime(2026, 6, 4)
df['posted_date'] = [base_date - timedelta(days=np.random.randint(1, 90)) for _ in range(n_jobs)]

# Add company description (for chatbot testing)
df['company_description'] = 'We are a technology company focused on data analytics and AI solutions.'

# Calculate average salary
df['salary_avg'] = (df['salary_min'] + df['salary_max']) / 2

# Save to CSV
df.to_csv(r'C:\Users\yelle\Job_Market_Analytics\data\raw\future_jobs_dataset.csv', index=False)

print("✅ Sample dataset created successfully!")
print(f"📊 Total jobs: {len(df)}")
print(f"📍 Location distribution:\n{df['location'].value_counts()}")
print(f"💼 Job type distribution:\n{df['job_type'].value_counts()}")
print(f"\n📁 File saved to: data/raw/future_jobs_raw.csv")

# Display first 5 rows
df.head()
```
